```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns 
from sklearn.cluster import KMeans
```

Drop the negative values or quantity

```python
data.drop(data[data['Quantity'] < 0].index, inplace=True)
```

```python
data.shape
```

Calculate of total amount or monrtary value

```python
data['sales'] = data['Quantity'] * data['UnitPrice']
new_data = data.groupby('CustomerID')['sales'].sum().reset_index()

new_data
```

```python
type(new_data)
```

Calculating the number of transaction of each customer


```python
num_trans = data.groupby('CustomerID')['InvoiceNo'].count().reset_index()

num_trans
```

```python
type(num_trans)
```

Calculate last transaction

```python
data['InvoiceDate'] = pd.to_datetime(data['InvoiceDate'])
data['InvoiceDate'].dtype
```

```python
data['InvoiceDate']
```

```python
data['Last Transaction'] = (data['InvoiceDate'].max() - data['InvoiceDate']).dt.days
data['Last Transaction']
```

```python
data.head(10)
```

```python
lt = data.groupby(['CustomerID', 'Country'])['Last Transaction'].max().reset_index()
```

```python
lt
```

```python
merge_table = pd.merge(lt, num_trans, how='inner', on='CustomerID')
new_df = pd.merge(merge_table, new_data, how='inner', on='CustomerID')
new_df
```

```python
plt.figure(figsize=(20,20))
sns.boxplot(data = new_df[['Last Transaction', 'InvoiceNo', 'sales']])
plt.show()
```

```python
IQR = new_df['sales'].quantile(0.75) - new_df['sales'].quantile(0.25)
lower_bound = new_df['sales'].quantile(0.25) - (IQR * 1.5)
upper_bound = new_df['sales'].quantile(0.75) + (IQR * 1.5)
new_df_iqr = new_df[(new_df['sales'] >= lower_bound) & (new_df['sales'] <= upper_bound)]
new_df_iqr.shape
```

```python
plt.figure(figsize=(20,20))
sns.boxplot(data = new_df_iqr[['Last Transaction', 'InvoiceNo', 'sales']])
plt.show()
```

```python
new_df_iqr.reset_index(drop=True, inplace=True)
new_df_iqr
```

```python
plt.figure(figsize = (20,20))
sns.heatmap(new_df_iqr.select_dtypes(include='number').corr(), annot=True, cmap='Greens')
plt.show()
```

```python
from sklearn.preprocessing import MinMaxScaler
```

```python
new2_df = new_df_iqr[['Last Transaction', 'InvoiceNo', 'sales']]
scaler = MinMaxScaler()
scaled_df = scaler.fit_transform(new2_df)
scaled_df = pd.DataFrame(scaled_df )
scaled_df.columns = ['Last Transaction' , 'InvoiceNo', 'sales']
scaled_df['country'] = new_df_iqr['Country']
scaled_df
```

K means clustering - Plot the graph using elbow method


```python
df_k = scaled_df.drop('country', axis=1)
# Elbow Method for k means
# import elbowvisualizer 

try:
    import yellowbrick.cluster  # type: ignore
except ModuleNotFoundError:
    %pip install -q yellowbrick
    import yellowbrick.cluster  # type: ignore
model = KMeans()
# k is range of number of clusters

visualizer = yellowbrick.cluster.KElbowVisualizer(model, k=(1,11),timing=False)
visualizer.fit(df_k)
visualizer.show()
```

```python
km = KMeans(n_clusters=3)
y_predicted = km.fit_predict(df_k)
df_k['cluster'] = y_predicted
df_k
```

```python
km.cluster_centers_
```

Model Evaluation

```python
km.labels_
```

```python
from sklearn.metrics import silhouette_samples, silhouette_score
score = silhouette_score(df_k , km.labels_, metric = 'euclidean')
print(score)
```

```python
from sklearn.cluster import AgglomerativeClustering

hierarchical = AgglomerativeClustering(n_clusters=3)

y_predicted_hierarchical = hierarchical.fit_predict(df_k)
df_k['cluster_hierarchical'] = y_predicted_hierarchical
df_k
```

```python
from sklearn.metrics import silhouette_samples, silhouette_score

silhouette_vals_hierarchical = silhouette_samples(df_k, y_predicted_hierarchical)
silhouette_score_hierarchical = silhouette_score(df_k, y_predicted_hierarchical)    
silhouette_score_hierarchical 
```

```python
from sklearn.cluster import DBSCAN

dbscan = DBSCAN(eps=0.2, min_samples=4)

y_predicted_dbscan = dbscan.fit_predict(df_k)
df_k['cluster_dbscan'] = y_predicted_dbscan
df_k 
```

```python
from sklearn.metrics import silhouette_samples, silhouette_score

silhouette_vals_dbscan = silhouette_samples(df_k, y_predicted_dbscan)
silhouette_score_dbscan = silhouette_score(df_k, y_predicted_dbscan)    
silhouette_score_dbscan 
```
