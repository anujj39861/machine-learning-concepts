```python
plt.figure(figsize=(10, 6))
sns.countplot(x="quality", data=data)
plt.title("Distribution of Wine Quality")
plt.xlabel("Quality")
plt.ylabel("Count")
print((data["quality"].value_counts(normalize=True)*100).sort_index().round(2))
```

```python
num_cols = [c for c in data.columns if c != "quality"]

data[num_cols].hist(figsize=(16,12), bins=25, edgecolor="black")
plt.suptitle("Feature Distributions", y=1.02)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(16,10))
for i, col in enumerate(num_cols, 1):
    plt.subplot(3, 4, i)
    sns.boxplot(y=data[col], color="skyblue")
    plt.title(col)
plt.tight_layout()
plt.show()
```

```python
def iqr_outlier_count(s):
    q1, q3 = s.quantile(0.25), s.quantile(0.75)
    iqr = q3 - q1
    low, high = q1 - 1.5*iqr, q3 + 1.5*iqr
    return ((s < low) | (s > high)).sum()

outlier_counts = pd.Series({c: iqr_outlier_count(data[c]) for c in num_cols}).sort_values(ascending=False)
print(outlier_counts)
```

```python
corr = data.corr(numeric_only=True)

plt.figure(figsize=(12,8))
sns.heatmap(corr, cmap="coolwarm", center=0)
plt.title("Correlation Heatmap")
plt.show()

print("Top correlations with quality:")
print(corr["quality"].drop("quality").sort_values(key=np.abs, ascending=False))
```

```python
top_feats = corr["quality"].drop("quality").abs().sort_values(ascending=False).head(4).index

for feat in top_feats:
    plt.figure(figsize=(7,4))
    sns.boxplot(x="quality", y=feat, data=data, palette="Set2")
    plt.title(f"{feat} vs quality")
    plt.show()
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVC
from imblearn.over_sampling import SMOTE
from imblearn.pipeline import Pipeline

X = data.drop("quality", axis=1)
y = data["quality"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, stratify=y, random_state=42
)

min_class_size = y_train.value_counts().min()
smote_k = max(1, min(5, min_class_size - 1))

pipe = Pipeline([
    ("scaler", StandardScaler()),
    ("smote", SMOTE(random_state=42, k_neighbors=smote_k)),
    ("svc", SVC(kernel="rbf"))
])

pipe.fit(X_train, y_train)
score = pipe.score(X_test, y_test)
print("Test accuracy:", score)
```
