```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt 
import seaborn as sns
```

```python
dataframe = pd.read_csv("WineQT.csv")
```

```python
dataframe.head()
```

```python
dataframe.columns
```

```python
set(dataframe["quality"])
```

```python
x = dataframe.drop("quality",axis =1)
y = dataframe["quality"]
```

```python
x.columns
```

```python
y.value_counts()
```

```python
from sklearn.model_selection import train_test_split
x_train ,x_test,y_train,y_test = train_test_split(x,y,test_size = 0.2,random_state = 42)
```

```python
x_train
```

```python
from sklearn.tree import DecisionTreeClassifier
model = DecisionTreeClassifier()
```

```python
model.fit(x_train,y_train)
```

```python
y_pred = model.predict(x_test)
```

```python
y_pred
```

```python
from sklearn import tree
plt.figure(figsize=(20,20))
tree.plot_tree(model,filled = True)
```

```python
from sklearn.metrics import r2_score
r2_score(y_test,y_pred)
```

```python
from sklearn.metrics import classification_report
print(classification_report(y_pred,y_test))
```

```python
from sklearn.ensemble import RandomForestClassifier , RandomForestRegressor

rf = RandomForestClassifier()
rf.fit(x_train,y_train)
```

```python
rf.score(x_test,y_test)
```

```python
y_predict_rf = rf.predict(x_test)
y_predict_rf
```

```python
rf.estimators_
```

# Hyperparameter tuning : *GridsearchCV and RandomsearchCV

```python
grid_param = {
    'n_estimators': [5, 10, 50, 100, 120, 150],
    'criterion': ['gini', 'entropy'],
    'max_depth': range(10),
    'min_samples_split': range(10)
}
```

```python
from sklearn.model_selection import GridSearchCV


grid_search_rf = GridSearchCV(estimator = rf , param_grid = grid_param , cv = 10 , n_jobs = 6 , verbose = 1)
```

```python
grid_search_rf.fit(x_train,y_train)
```

```python
grid_search_rf.best_params_
```

```python
random_param={
    'n_estimators': [5, 10, 50, 100, 120, 150],
    'criterion': ['gini', 'entropy'],
    'max_depth': range(10),
    'min_samples_split': range(10)
}
```

```python
from sklearn.model_selection import RandomizedSearchCV

random_search_rf = RandomizedSearchCV(estimator = rf , param_distributions = random_param , cv = 10 , n_jobs = 6 , verbose = 1)
```

```python
random_search_rf.fit(x_train,y_train)
```

```python
random_search_rf.best_params_
```

```python
rf_new = RandomForestClassifier(criterion = 'gini' , max_depth = 9 , min_samples_split = 2, n_estimators = 150)
rf_new.fit(x_train,y_train)
```

```python
rf_new.score(x_test,y_test)
```
